### Correcting `fillna` Usage

As discussed, the `None` values in your DataFrame were introduced because of an incorrect assignment when using `fillna(..., inplace=True)`. When `inplace=True` is set, the method modifies the DataFrame directly and returns `None`. If you then assign this `None` back to the column, the column becomes entirely `None`.

There are two correct ways to use `fillna`:

1.  **Modify in place without re-assignment:**
    ```python
    df['column_name'].fillna(df['column_name'].mean(), inplace=True)
    ```
    
2.  **Assign the result without `inplace=True` (preferred as it's more explicit and avoids `FutureWarning`):**
    ```python
    df['column_name'] = df['column_name'].fillna(df['column_name'].mean())
    ```

Let's restart with a fresh load of your data and apply the second (and generally recommended) method.

````python
# Reload the original dataset to ensure a clean state
df = pd.read_csv('Life Expectancy Data - Life Expectancy Data.csv')

# Display initial nulls for 'Life expectancy' before correction
print("Nulls in 'Life expectancy' before correction:")
print(df['Life expectancy'].isnull().sum())

# Correctly fill missing values for 'Life expectancy' by assigning the result
df['Life expectancy'] = df['Life expectancy'].fillna(df['Life expectancy'].mean())

# Verify that null values are gone for 'Life expectancy'
print("\nNulls in 'Life expectancy' after correction:")
print(df['Life expectancy'].isnull().sum())

# Apply the same correction pattern to other relevant columns
df['Total expenditure'] = df['Total expenditure'].fillna(df['Total expenditure'].mean())
df['Adult Mortality'] = df['Adult Mortality'].fillna(df['Adult Mortality'].mean())
df['Alcohol'] = df['Alcohol'].fillna(df['Alcohol'].mean())
df['Hepatitis B'] = df['Hepatitis B'].fillna(df['Hepatitis B'].mean())
df['BMI'] = df['BMI'].fillna(df['BMI'].mean())
df['Polio'] = df['Polio'].fillna(df['Polio'].mean())
df['Diphtheria'] = df['Diphtheria'].fillna(df['Diphtheria'].mean())
df['GDP'] = df['GDP'].fillna(df['GDP'].mean())
df['Population'] = df['Population'].fillna(df['Population'].mean())
df['thinness 5-9 years'] = df['thinness 5-9 years'].fillna(df['thinness 5-9 years'].mean())
df['thinness  1-19 years'] = df['thinness  1-19 years'].fillna(df['thinness  1-19 years'].mean())
df['Schooling'] = df['Schooling'].fillna(df['Schooling'].mean())
df['Income composition of resources'] = df['Income composition of resources'].fillna(df['Income composition of resources'].mean())

print("\nAll nulls after correcting all columns:")
print(df.isnull().sum())
````

````python
# Label Encoder
le = LabelEncoder()
df['Country'] = le.fit_transform(df['Country'])
df['Status'] = le.fit_transform(df['Status'])
df.head(5)
````

````python
# Input and Output
x = df.drop(columns="Life expectancy")
y = df['Life expectancy']
````

````python
#Train Test split
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
print(x_train.head())
print(x_test.head())
````

````python
#Standard Scaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.fit_transform(x_test)
````

````python
# ANN Building
ann = Sequential()

ann.add(Dense(units=320,activation='relu'))
ann.add(Dense(units=180,activation='relu'))
ann.add(Dense(units=90,activation='relu'))
ann.add(Dense(units=19,activation='relu'))

ann.add(Dense(units=1,activation='relu'))
````

````python
ann.compile(optimizer='adadelta',loss='mean_squared_error',metrics=['mae'])
````

````python
from tensorflow.keras.callbacks import EarlyStopping
````

````python
early_stopping = EarlyStopping(monitor='val_loss',patience=10,restore_best_weights=True)
````

````python
history = ann.fit(x_train,y_train,validation_split=0.2,batch_size=320,epochs=100,callbacks=[early_stopping])
````

````python
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error
````

````python
y_pred = ann.predict(x_test)
````

````python
mse = mean_squared_error(y_test,y_pred)
r2 = r2_score(y_test,y_pred)
print("R2_score : ",r2)
print("MSE : ",mse)
````
